```python
from langchain_core.tools import tool
@tool
def recommend_restaurant(city: str) -> str:
    """Recommends a restaurant for a given city."""
    restaurants = {
        "Lahore": "Cuckoo's Den - rooftop dining near Badshahi Mosque",
        "Karachi": "Kolachi - seaside dining at Do Darya",
        "Islamabad": "Monal - hilltop restaurant with city views",
    }
    return restaurants.get(city, "No restaurant recommendation available for this city")
```

```python
tools =  [
    get_weather,
    calculator,
    recommend_restaurant
]
```

```python
llm_with_tools = llm.bind_tools(tools)
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful travel assistant."),
    ("placeholder","{chat_history}"),
    ("human", "{input}"),
    ("placeholder", "{agent_scratchpad}")
]
)
agnet = create_tool_calling_agent(
    llm=llm,
    tools=tools,
    prompt=prompt
)
agent_executor = AgentExecutor.from_agent_and_tools(
    agent=agent,
    tools=tools,
    verbose=True
)
```

```python
class TravelMemory:
    def __init__(self):
        self.chat_history = []
        self.last_city = None

    def add(self, human_msg, ai_msg, city=None):
        self.chat_history.append(("human", human_msg))
        self.chat_history.append(("ai", ai_msg))
        self.chat_history = self.chat_history[-10:]
        if city:
            self.last_city = city

    def get_history(self):
        return self.chat_history

memory = TravelMemory()
known_cities = ["Lahore", "Karachi", "Islamabad"]

def ask(user_input):
    for city in known_cities:
        if city.lower() in user_input.lower():
            memory.last_city = city

    context_input = user_input
    if memory.last_city and memory.last_city.lower() not in user_input.lower():
        context_input = f"(Context: the user's city is {memory.last_city}) {user_input}"

    result = agent_executor.invoke({
        "input": context_input,
        "chat_history": memory.get_history()
    })

    memory.add(user_input, result["output"], memory.last_city)
    return result["output"]
```

```python
print(ask("I'm visiting Lahore tomorrow. What's the weather?"))
print(ask("Recommend a restaurant."))
print(ask("How much will dinner cost for 4 people if each meal costs $25?"))
```
